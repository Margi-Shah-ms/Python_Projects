```python
import numpy as np

class DataAnalytics:
    def __init__(self):
        self.__arr = None      

    def get_array(self):
        return self.__arr

    def set_array(self, a2):
        self.__arr = a2

    @staticmethod
    def inputs(input_s):
        arr = []
        for x in input_s.split():
            arr.append(int(x))
        return arr

    def add(self, a2):
        return self.__arr + a2
    def subtract(self, a2):
        return self.__arr - a2
    def multiply(self, a2):
        return self.__arr * a2
    def divide(self, a2):
        return self.__arr / a2
    def dot_product(self, a2):
        return np.dot(self.__arr, a2)
    def matrix_multiply(self, a2):
        return np.matmul(self.__arr, a2)
    
    def combine(self, a2):
        return np.vstack((self.__arr, a2))
    def vsplit(self, sp_lit):
        return np.vsplit(self.__arr, sp_lit)
    def hsplit(self, sp_lit):
        return np.hsplit(self.__arr, sp_lit)
    def search(self, value):
        return np.where(self.__arr == value)
    def sort(self):
        return np.sort(self.__arr)
    def filter_data(self, c, val):
        if c == ">":
            return self.__arr[self.__arr > val]
        elif c == "<":
            return self.__arr[self.__arr < val]
        elif c == "==":
            return self.__arr[self.__arr == val]
        elif c == ">=":
            return self.__arr[self.__arr >= val]
        elif c == "<=":
            return self.__arr[self.__arr <= val]

    def get_sum(self):
        return np.sum(self.__arr)
    def get_mean(self):
        return np.mean(self.__arr)
    def get_median(self):
        return np.median(self.__arr)
    def get_std(self):
        return np.std(self.__arr)
    def get_var(self):
        return np.var(self.__arr)
    def get_min(self):
        return np.min(self.__arr)
    def get_max(self):
        return np.max(self.__arr)
    def get_percentile(self, p):
        return np.percentile(self.__arr, p)


DA = DataAnalytics()
print("Welcome to the Numpy Analyzer!")
print("=====================================")
while True:
    print("\nChoose an option")
    print("1. Create a Numpy Array")
    print("2. Perform Mathematical Operations")
    print("3. Combine or Split Arrays")
    print("4. Search, Sort, or Filter Arrays")
    print("5. Compute Aggregates and Statistics")
    print("6. Exit")
    user = input("\nEnter your choice: ")   
    match user:
        case "1":
            print("Select the type of array to create:")
            print("1. 1D Array")
            print("2. 2D Array")
            print("3. 3D Array")
            choice = input("Enter your choice: ")  
            match choice:
                case "1":
                    a = int(input("Enter the size: "))
                    c = input(f"Enter {a} elements for the array separated by space: ")
                    arr = DA.inputs(c)
                    DA.set_array(np.array(arr))
                    print("\nArray created successfully:")
                    new = DA.get_array()
                    print(new)

                case "2":
                    a = int(input("Enter the number of rows: "))
                    b = int(input("Enter the number of columns: "))
                    d = a * b
                    c = input(f"Enter {d} elements for the array separated by space: ")
                    arr = DA.inputs(c)
                    DA.set_array(np.array(arr).reshape(a, b))
                    print("\nArray created successfully:")
                    new = DA.get_array()
                    print(new)

                case "3":
                    a = int(input("Enter the dimension: "))
                    b = int(input("Enter the number of rows: "))
                    c = int(input("Enter the number of columns: "))
                    d = a * b * c
                    e = input(f"Enter {d} elements separated by space: ")
                    data = DA.inputs(e)
                    DA.set_array(np.array(data).reshape(a, b, c))
                    print("\nArray created successfully:")
                    new = DA.get_array()
                    print(new)

            if choice == "2" or choice == "3":
                print("\nChoose an operation:")
                print("1. Indexing")
                print("2. Slicing")
                print("3. Go Back")
                sub_choice = input("Enter your choice: ")
                match sub_choice:
                    case "1":
                        r = int(input("Enter row index: "))
                        col = int(input("Enter column index: "))
                        print("\nElement at position:", new[r, col])

                    case "2":
                        row = input("Enter the row range (start:end): ").split(":")
                        colum = input("Enter the column range (start:end): ").split(":")
                        row_s, row_e = int(row[0]), int(row[1])
                        col_s, col_e = int(colum[0]), int(colum[1])
                        print("\nSliced Array:")
                        print(new[row_s:row_e, col_s:col_e])   

                    case "3":
                        print("Back to main menu.")

        case "2":
            new = DA.get_array()
            if new is None:
                print("\nPlease create an array first!")
                continue
            print("\nChoose a mathematical operation:")
            print("1. Addition")
            print("2. Subtraction")
            print("3. Multiplication")
            print("4. Division")
            choice = input("\nEnter your choice: ")
            d = new.size
            c = input(f"\nEnter same-size array elements ({d} elements separated by space): ")
            arr = DA.inputs(c)
            arr = np.array(arr).reshape(new.shape)
            print("\nOriginal Array:")
            print(new)
            print("\nSecond Array:")
            print(arr)  
            match choice:
                case "1":
                    print("\nResult of Addition:")
                    print(DA.add(arr))
                case "2":
                    print("\nResult of Subtraction:")
                    print(DA.subtract(arr))
                case "3":
                    print("\nResult of Multiplication:")
                    print(DA.multiply(arr))
                case "4":
                    if 0 in arr:
                        print("\nError: Division by zero is not allowed!")
                    else:
                        print("\nResult of Division:")
                        print(DA.divide(arr))
                case _:
                    print("\nInvalid choice!")

        case "3":
            new = DA.get_array()
            if new is None:
                print("\nPlease create an array first!")
                continue
            print("Choose an option:")
            print("1. Combine Arrays")
            print("2. Split Array")
            c_choice = input("Enter your choice: ")
            match c_choice:
                case "1":
                    d = new.size
                    c = input(f"\nEnter the elements of another array to combine ({d} elements separated by space): ")
                    arr2 = DA.inputs(c)
                    arr2 = np.array(arr2).reshape(new.shape)
                    print("\nOriginal Array:")
                    print(new)
                    print("\nSecond Array:")
                    print(arr2)
                    print("\nCombined Array (Vertical Stack):")
                    print(DA.combine(arr2))
                case "2":
                    print("\nChoose Split Type:")
                    print("1. Vertical Split (vsplit)")
                    print("2. Horizontal Split (hsplit)")
                    choice = input("Enter your choice: ")
                    parts = int(input("Enter sections to split: "))
                    if choice == "1":
                        p_l = DA.vsplit(parts)
                    else:
                        p_l = DA.hsplit(parts)

                    print("\nSplit Result:")
                    for i in p_l:
                        print(i)
                case _:
                    print("\nInvalid choice!")

        case "4":
            new = DA.get_array()
            if new is None:
                print("\nPlease create an array first!")
                continue
            print("\nChoose an option:")
            print("1. Search a value")
            print("2. Sort the array")
            print("3. Filter values")
            s_choice = input("Enter your choice: ")
            if s_choice == "1":
                val = int(input("Enter value to search: "))
                print("Value found at index:", DA.search(val))
            elif s_choice == "2":
                print("\nOriginal Array:")
                print(new)
                print("\nSorted Array:")
                print(DA.sort())
                print("(Sorting applied row-wise.)")
            elif s_choice == "3":
                op = input("Enter condition operator(>, <, ==, >=, <=):")
                val = int(input("Enter thresh value: "))
                print("\nFiltered Elements:")
                print(DA.filter_data(op, val))

        case "5":
            new = DA.get_array()
            if new is None:
                print("\nPlease create an array first!")
                continue
            print("\nChoose an aggregate/statistical operation:")
            print("1. Sum")
            print("2. Mean")
            print("3. Median")
            print("4. Standard Deviation")
            print("5. Variance")
            choice = input("Enter your choice: ")
            print("\nOriginal Array:")
            print(new)
            if choice == "1":
                print("\nSum of Array:", DA.get_sum())
            elif choice == "2":
                print("\nMean of Array:", DA.get_mean())
            elif choice == "3":
                print("\nMedian of Array:", DA.get_median())
            elif choice == "4":
                print("\nStandard Deviation of Array:", DA.get_std())
            elif choice == "5":
                print("\nVariance of Array:", DA.get_var())
            else:
                print("\nInvalid Choice.")

        case "6":
            print("\nThank you for using the NumPy Analyzer! Goodbye!")
            break

        case _:
            print("Invalid Choice. Please enter a valid choice.")
```
