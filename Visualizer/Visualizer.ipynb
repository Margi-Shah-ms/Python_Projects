```python
import pandas as pd
import numpy as np
import matplotlib.pyplot as plt
import seaborn as sns
import os
```

```python
class SalesDataAnalyzer:
    def __init__(self, file_path=None):
        self.data = None
        self.last_fig = None
        if file_path:
            self.load_data(file_path)
            
    def __del__(self):
        pass
    
    def load_data(self, file_path):
        if os.path.exists(file_path):
            self.data = pd.read_csv(file_path)
            print("Dataset loaded successfully!")
        else:
            print(f"Error: File '{file_path}' not found.")

    def show_head(self):
        if self.data is None:
            print("Load dataset first!")
            return 
        print(self.data.head())

    def show_tail(self):
        if self.data is None:
            return print("Load dataset first!")
        print(self.data.tail())

    def show_columns(self):
        if self.data is None:
            return print("Load dataset first!")
        print(self.data.columns)

    def show_dtypes(self):
        if self.data is None:
            return print("Load dataset first!")
        print(self.data.dtypes)

    def show_info(self):
        if self.data is None:
            return print("Load dataset first!")

        print("\n== Basic Information of Dataset ==")
        print("Number of rows:", self.data.shape[0])
        print("Number of columns:", self.data.shape[1])
        print("Column names:", self.data.columns.tolist())

        print("\nMissing Values:")
        print(self.data.isnull().sum())

        print("\nDataset Info:")
        self.data.info()

    def show_missing(self):
        if self.data is None:
            print("Load dataset first!")
            return
        null_rows = self.data[self.data.isnull().any(axis=1)]
        if not null_rows.empty:
            print(null_rows)

    def fill_missing_mean(self):
        if self.data is None:
            print("Load dataset first!")
            return
        numeric_cols = self.data.select_dtypes(include=[np.number]).columns
        self.data[numeric_cols] = (self.data[numeric_cols].fillna(self.data[numeric_cols].mean()))
        print("Missing numeric values filled with column mean.")

    def drop_missing(self):
        if self.data is None:
            print("Load dataset first!")
            return 
        self.data.dropna(inplace=True)
        print("Rows with missing values dropped.")

    def replace_missing_val(self, val):
        if self.data is None:
            print("Load dataset first!")
            return 
        self.data.fillna(val, inplace=True)
        print(f"Missing values replaced with '{val}'.")


    def mathematical_operations(self):
        if self.data is None: 
            print("Load dataset first!")
            return 
        numeric_cols = self.data.select_dtypes(include=[np.number]).columns.tolist()
        if not numeric_cols: 
            print("No numeric columns available.")
            return 
        col = numeric_cols[0]
        arr = self.data[col].to_numpy()
        print(f"NumPy Array from '{col}':", arr[:5])
        print("Array Slicing (first 3):", arr[:3])
        print("Element-wise x 2:", arr[:5] * 2)

    def statistical_analysis(self):
        if self.data is not None:
            print(self.data.describe()) 
        else:
            print("Load dataset first!")
            
    def create_plot(self, plot_type):
        if self.data is None: 
            print("Load dataset first!")
            return 
        
        fig, ax = plt.subplots(figsize=(8, 5))
        
        match plot_type:
            case 'bar':
                print("\n== Bar Plot ==")
                x = input("Enter x-axis column name: ")
                y = input("Enter y-axis column name: ")
                print("Generating bar plot...")
                sns.barplot(data=self.data, x=x, y=y, ax=ax)
                
            case 'line':
                print("\n== Line Plot ==")
                x = input("Enter x-axis column name: ")
                y = input("Enter y-axis column name: ")
                print("Generating line plot...")
                sns.lineplot(data=self.data, x=x, y=y, ax=ax)
                
            case 'scatter':
                print("\n== Scatter Plot ==")
                x = input("Enter x-axis column name: ")
                y = input("Enter y-axis column name: ")
                print("Generating scatter plot...")
                sns.scatterplot(data=self.data, x=x, y=y, ax=ax)
                
            case 'pie':
                print("\n== Pie Chart ==")
                col = input("Enter column name for Pie Chart: ")
                print("Generating pie chart...")
                self.data[col].value_counts().plot.pie(autopct='%1.1f%%', ax=ax)
                
            case 'hist':
                print("\n== Histogram ==")
                col = input("Enter column name for Histogram: ")
                print("Generating histogram...")
                sns.histplot(self.data[col], kde=True, ax=ax)
                
            case 'stack':
                print("\n== Stack Plot ==")
                y = input("Enter y-axis column name: ")
                print("Generating stack plot...")
                ax.stackplot(range(len(self.data)), self.data[y])
        plt.tight_layout()
        plt.show()
        self.last_fig = fig
        print(f"{plot_type.capitalize()} plot displayed successfully!")

    def save_visualization(self):
            if self.last_fig is None: 
                print("No plot available to save!")
                return 
            filename = input("Enter file name to save (e.g., plot.png): ")
            self.last_fig.savefig(filename)
            print(f"Visualization saved as {filename} successfully!")
```

```python
analyzer = SalesDataAnalyzer()
```

```python
while True:
    print("\n======== Data Analysis & Visualization Program ========")
    print("Please select an option:")
    print("1. Load Dataset")
    print("2. Explore Data")
    print("3. Perform DataFrame Operations")
    print("4. Handle Missing Data")
    print("5. Generate Descriptive Statistics")
    print("6. Data Visualization")
    print("7. Save Visualization")
    print("8. Exit")
    print("=======================================================")
    choice = input("\nEnter your choice: ")
    match choice:
        
        case '1':
            print("\n== Load Dataset ==")
            path = input("Enter the path of the dataset (CSV file): ")
            analyzer.load_data(path)

        case '2':
            while True:
                print("\n== Explore Data ==")
                print("1. Display the first 5 rows")
                print("2. Display the last 5 rows")
                print("3. Display column names")
                print("4. Display data types")
                print("5. Display basic info")
                choice = input("Enter your choice: ")
                match choice:
                    case '1': 
                        analyzer.show_head()
                        break
                    case '2': 
                        analyzer.show_tail()
                        break
                    case '3': 
                        analyzer.show_columns()
                        break
                    case '4': 
                        analyzer.show_dtypes()
                        break
                    case '5': 
                        analyzer.show_info()
                        break
                    case '6': 
                        break
                    case _: 
                        break

        case '3':
            analyzer.mathematical_operations()

        case '4':
            while True:
                print("\n== Handling Missing Data ==")
                print("1. Display rows with missing values")
                print("2. Fill missing values with mean")
                print("3. Drop rows with missing values")
                print("4. Replace missing values with a specific value")
                choice = input("Enter your choice: ")
                match choice:
                    case '1': 
                        analyzer.show_missing()
                        print("\nNo missing values found in the dataset!")
                        break
                    case '2': 
                        analyzer.fill_missing_mean()
                        break
                    case '3': 
                        analyzer.drop_missing()
                        break
                    case '4': 
                        val = input("Enter replacement value: ")
                        analyzer.replace_missing_val(val)
                        break
                    case _: 
                        break

        case '5':
            analyzer.statistical_analysis()

        case '6':
            while True:
                print("\n== Data Visualization ==")
                print("1. Bar Plot")
                print("2. Line Plot")
                print("3. Scatter Plot")
                print("4. Pie Chart")
                print("5. Histogram")
                print("6. Stack Plot")
                choice = input("Enter your choice: ")
                match choice:
                    case '1': 
                        analyzer.create_plot('bar')
                        break
                    case '2': 
                        analyzer.create_plot('line')
                        break
                    case '3': 
                        analyzer.create_plot('scatter')
                        break
                    case '4': 
                        analyzer.create_plot('pie')
                        break
                    case '5': 
                        analyzer.create_plot('hist')
                        break
                    case '6': 
                        analyzer.create_plot('stack')
                        break
                    case _: 
                        break

        case '7':
            analyzer.save_visualization()

        case '8':
            print("\nExiting program. Goodbye!")
            break

        case _:
            print("Invalid Option, Please try again.")
```
